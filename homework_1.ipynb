## Задание 2. Подготовка данных к анализу

Нужно создать новый список `clean_orders`, не изменяя `orders`:
- у имени клиента убрать пробелы по краям;
- город очистить от пробелов и привести к нижнему регистру;
- сумму преобразовать в `float`, а если это невозможно — записать `None`.

Сначала опишем функции очистки, затем применим их ко всем заказам.

```python
def clean_customer(name):
    """Убирает лишние пробелы по краям имени клиента."""
    return name.strip()


def clean_city(city):
    """Убирает пробелы по краям и приводит название города к нижнему регистру."""
    return city.strip().lower()


def parse_amount(value):
    """Преобразует сумму в float. Если преобразовать нельзя, возвращает None."""
    try:
        return float(value.strip())
    except (ValueError, AttributeError):
        # ValueError — строку нельзя превратить в число ("нет данных", "")
        # AttributeError — на вход пришла не строка (например, None)
        return None
```

```python
def clean_order(order):
    """Возвращает новый словарь с очищенными полями заказа. Исходный заказ не меняется."""
    return {
        "order_id": order["order_id"],
        "customer": clean_customer(order["customer"]),
        "city": clean_city(order["city"]),
        "amount": parse_amount(order["amount"]),
        "status": order["status"],
        "email": order["email"],
    }
```

```python
clean_orders = []
for order in orders:
    clean_orders.append(clean_order(order))

for order in clean_orders:
    print(order)
```

Убедимся, что исходный список не изменился: сравним первый исходный заказ с очищенным.

```python
print("Исходный заказ: ", orders[0])
print("Очищенный заказ:", clean_orders[0])
```

## Задание 3. Анализ очищенных данных

Все вычисления выполняются по списку `clean_orders`.

### 3.1–3.2. Количество заказов со статусами `paid` и `cancelled`

```python
paid_count = 0
cancelled_count = 0

for order in clean_orders:
    if order["status"] == "paid":
        paid_count += 1
    elif order["status"] == "cancelled":
        cancelled_count += 1

print("Заказов со статусом paid:", paid_count)
print("Заказов со статусом cancelled:", cancelled_count)
```

### 3.3–3.4. Общая и средняя сумма оплаченных заказов

Учитываем только заказы со статусом `paid`, у которых сумма известна (не `None`).

```python
paid_total = 0.0
paid_with_amount_count = 0

for order in clean_orders:
    if order["status"] == "paid" and order["amount"] is not None:
        paid_total += order["amount"]
        paid_with_amount_count += 1

if paid_with_amount_count > 0:
    paid_average = paid_total / paid_with_amount_count
else:
    paid_average = None

print("Оплаченных заказов с корректной суммой:", paid_with_amount_count)
print("Общая сумма оплаченных заказов:", round(paid_total, 2))
print("Средняя сумма оплаченного заказа:", round(paid_average, 2))
```

### 3.5. Множество уникальных городов

```python
unique_cities = set()
for order in clean_orders:
    unique_cities.add(order["city"])

print("Уникальные города:", unique_cities)
```

### 3.6. Количество заказов по каждому городу

```python
orders_by_city = {}
for order in clean_orders:
    city = order["city"]
    if city in orders_by_city:
        orders_by_city[city] += 1
    else:
        orders_by_city[city] = 1

print("Количество заказов по городам:", orders_by_city)
```

### 3.7. Клиенты из Москвы с оплаченным заказом на сумму больше 1000

Условия: город `москва`, статус `paid`, сумма известна и больше 1000.

```python
moscow_big_customers = []
for order in clean_orders:
    if (
        order["city"] == "москва"
        and order["status"] == "paid"
        and order["amount"] is not None
        and order["amount"] > 1000
    ):
        moscow_big_customers.append(order["customer"])

print("Клиенты из Москвы (paid, сумма > 1000):", moscow_big_customers)
```

## Задание 4. Проверка решения

Проверим функции очистки на дополнительных значениях. Для удобства напишем вспомогательную функцию `check`, которая вызывает функцию очистки, сравнивает результат с ожидаемым и печатает `OK` или `ОШИБКА`.

```python
def check(func, value, expected):
    """Вызывает func(value) и сравнивает результат с ожидаемым значением."""
    result = func(value)
    if result == expected:
        verdict = "OK"
    else:
        verdict = "ОШИБКА"
    print(verdict, "|", func.__name__, "(", repr(value), ") ->", repr(result), "| ожидалось:", repr(expected))
```

### 4.1. Проверка городов

```python
check(clean_city, "  МОСКВА ", "москва")
check(clean_city, "Казань", "казань")
check(clean_city, "санкт-Петербург ", "санкт-петербург")
```

### 4.2. Проверка сумм: корректные и некорректные значения

```python
check(parse_amount, "1500", 1500.0)
check(parse_amount, "99.90", 99.9)
check(parse_amount, "нет данных", None)
check(parse_amount, "", None)
```

### 4.3. Собственные проверки

- сумма с пробелами по краям (`" 450 "`) должна корректно превращаться в число;
- строка, где число смешано с буквами (`"12abc"`), не должна считаться суммой;
- если суммы нет совсем (`None` вместо строки), функция не должна падать;
- имя клиента: убираются пробелы только по краям, пробел между именем и фамилией сохраняется.

```python
check(parse_amount, " 450 ", 450.0)
check(parse_amount, "12abc", None)
check(parse_amount, None, None)
check(clean_customer, "   Анна   Смирнова  ", "Анна   Смирнова")
```
